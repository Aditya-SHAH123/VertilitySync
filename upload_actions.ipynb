# Upload Actions Notebook
This notebook records the shell commands used to upload the project to GitHub and includes a small helper to inspect git status locally.

```python
# Initialize and push project
git init
git add .
git commit -m "Initial import from local workspace"
git branch -M main
git remote add origin https://github.com/Aditya-SHAH123/VertilitySync.git
git push -u origin main

# Add README and push
git add README.md
git commit -m "Add README.md"
git push

# Stage and push later updates
git add -A
git commit -m "Update modified files"
git push
```

```python
import subprocess
status = subprocess.run(["git", "status", "--porcelain"], capture_output=True, text=True)
print("Git status (porcelain):")
print(status.stdout)
diff = subprocess.run(["git", "diff", "--name-only"], capture_output=True, text=True)
print("Changed files:")
print(diff.stdout)
```
